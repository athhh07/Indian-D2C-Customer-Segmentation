# Notebook 03 – Data Integration

## Objective

### Merge multiple relational datasets into a single customer-level dataset.

### Output: master_customer.csv

```python
import pandas as pd
from pathlib import Path
PROCESSED = Path("data/processed")
```

```python
customers= pd.read_csv(PROCESSED/"customers.csv")
orders= pd.read_csv(PROCESSED/"orders.csv")
order_items= pd.read_csv(PROCESSED/"order_items.csv")
website_sessions= pd.read_csv(PROCESSED/"website_sessions.csv")
```

### Merging Customers & Orders:

```python
customer_orders = customers.merge(
    orders,
    on="customer_id",
    how="left"
)
customer_orders.shape
```

### Merging order items:

```python
customer_orders = customer_orders.merge(
    order_items,
    on="order_id",
    how="left"
)
customer_orders.shape
```

### Merging website sessions:

```python
master = customer_orders.merge(
    website_sessions,
    on="customer_id",
    how="left"
)
master.shape
```

### Validating Operations:

```python
print("Customers:",customers.customer_id.nunique())
print("Master:",master.customer_id.nunique())
```

```python
master.to_csv(
    PROCESSED/"master_customer.csv",
    index=False
)
print("Master dataset saved..!")
```
